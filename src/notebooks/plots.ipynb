```python
import matplotlib.pyplot as plt
import numpy as np

locations=np.array(dataset["locations"])
# Extract x and y coordinates
x = locations[:, 0]*1.00  # First column
y = locations[:, 1]*1.00  # Second column
locations[:, 0]=x
locations[:, 1]=y
```

```python
# create  bounding lines
trace_stds=np.array(dataset["log_stds"])[:,0]
```

```python
unit_vectors=np.array(dataset["unit_vectors"])
# locations=np.array(locations)
```

```python
# points=[]
# for unit_vec,trace_std,location in zip(dataset["unit_vectors"],trace_stds,locations):
#     points.append(np.array(unit_vec)*trace_std+location)
# points=np.array(points)

# magnify 
points_1=10*(trace_stds[...,None]*unit_vectors)+locations 
points_2=10*(trace_stds[...,None]*(-1)*unit_vectors)+locations
# points=np.vstack([points_1,points_2])
```

```python
# Create the scatter plot

fontsize=18
plt.rcParams["font.size"]=fontsize
plt.rcParams['axes.labelsize'] = fontsize
plt.rcParams['axes.titlesize'] = fontsize
plt.style.use('seaborn-v0_8-darkgrid')
# sns.set_context("paper", font_scale=1.2)

plt.figure(figsize=(8, 6))

# Get the x and y limits of your data
x_min, x_max = min(min(x), min(points_1[:,0]), min(points_2[:,0])), max(max(x), max(points_1[:,0]), max(points_2[:,0]))
y_min, y_max = min(min(y), min(points_1[:,1]), min(points_2[:,1])), max(max(y), max(points_1[:,1]), max(points_2[:,1]))

# Add a small margin
x_margin = -0.00
y_margin=-0.00
x_range = x_max - x_min
y_range = y_max - y_min
x_min -=x_margin * x_range
x_max += x_margin * x_range
y_min -= y_margin * y_range
y_max += y_margin * y_range

# Display the image with extent matching the data range
im = plt.imread('town01aerial.jpg')
plt.imshow(im, extent=[x_min, x_max, y_min, y_max], zorder=1)

# Plot the data with smaller scatter points
# plt.scatter(x, y, s=10, color='blue')  # Blue scatter points
# plt.xlabel('X Position')
# plt.ylabel('Y Position')
# plt.title('X-Y Position Plot')
plt.grid(False, linewidth=0.5)  # Thinner grid lines

# Add optional features with thinner lines and different colors
plt.plot(x, y, linewidth=0.5, zorder=2, color='red',label="Trajectory")  # Red line
plt.plot(points_1[:,0], points_1[:,1], linewidth=0.5, zorder=2,label="Uncertainty", color='indigo')  # Green line
plt.plot(points_2[:,0], points_2[:,1], linewidth=0.5, zorder=2, color='black')  # Yellow line
plt.tight_layout()

# Set the plot limits explicitly
# plt.xlim(x_min, x_max)
# plt.ylim(y_min, y_max)
# Display the plot
# plt.legend(loc='upper left', borderaxespad=0.,
#           frameon=True, fancybox=True, shadow=True,fontsize=fontsize)
plt.axis('off')
plt.savefig('plot.pdf')
plt.show()
```
