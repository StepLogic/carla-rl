```python
with open('/home/kojogyaase/Projects/Research/carla-rl/plot_data.pickle', 'rb') as handle:
    dataset=pickle.load(handle)
```

```python
# dataset[]
```

```python
import matplotlib.pyplot as plt
import numpy as np

locations=np.array(dataset["locations"])
# Extract x and y coordinates
x = locations[:, 0]  # First column
y = locations[:, 1]  # Second column


# create  bounding lines
trace_stds=np.array(dataset["trace_log_stds"])
points=[]
for unit_vec,trace_std,location in zip(dataset["unit_vectors"],trace_stds,locations):
    points.append(np.array(unit_vec)*trace_std+location)
points=np.array(points)
```

```python
# Create the scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(x, y)
plt.xlabel('X Position')
plt.ylabel('Y Position')
plt.title('X-Y Position Plot')
plt.grid(True)
# Add optional features
plt.scatter(x, y, c=locations[:, 2])  # Color points by Z value
plt.scatter(points[:,0], points[:,1], c=locations[:, 2])  # Color points by Z value

# Display the plot
plt.show()
```
